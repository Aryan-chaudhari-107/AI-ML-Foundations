## Why Do We Need Serialization?

Pydantic objects are Python objects.

But APIs and databases usually understand:

- JSON
- dictionaries

So we convert model data into serializable format.

```python
from pydantic import BaseModel

#step 1 : model
class Address(BaseModel):
    city: str
    state: str
    pin: int

class Patient(BaseModel):

    name: str
    gender: str
    age : int
    address: Address # address is complex data 


def detail(patient: Patient):
    print(patient.name)
    print(patient.gender)
    print(patient.age)
    print(patient.address)
    

    
# step 2 : dict

address_dict = {'city':'amd', 'state':'guj', 'pin': 646146}
address1 = Address(**address_dict)

patient_dict = {'name': 'Aryan', 'gender':'male', 'age': 22, 'address': address1}
patient1 = Patient(**patient_dict)


temp = patient1.mode_dump() # covert to python dict
print(temp)
print(type(temp))

temp2 = patient1.mode_dump_jason()
print(temp2)
print(type(temp2)) 
```

## INCLUDE AND EXCLUDE

```python
temp = patient1.mode_dump(include=['name', 'gender'])

temp = patient1.mode_dump(exclude=['name', 'gender'])
```

## FOR NESTED

```python
temp = patient1.mode_dump(exclude = {'address' : 'state'})
```
