## Computed Fields in Pydantic

- Computed fields are fields whose values are calculated automatically from other fields.

- You do not pass them manually.

```python
from pydantic import BaseModel, EmailStr, AnyUrl, computed_field
from typing import List,Dict, Optional, Annotated

# step 1: model
class Patient(BaseModel):

    name: str
    email: EmailStr
    age: int
    weight: float #kg
    height: float #mtr
    
    #-------------------------------------------------------------------
    # Now we want a BMI field, but we don't want the user to provide it manually.
    # Otherwise, the user would have to calculate the BMI themselves.
    #-------------------------------------------------------------------
    @computed_field
    @property
    def bmi(self) -> float:
        bmi = round(self.weight / (self.height * 2), 2)
        return bmi
    #-------------------------------------------------------------------

    
def insert_patient_data(patient: Patient):
    print(patient.name)
    print(patient.email)
    print(patient.age)
    print(patient.weight)
    print(patient.height)
    print('BMI', patient.bmi)
    
    print("Inserted into database")




#step 2: pass raw input data (created dict)

patient_info = {'name': 'aryan', 'email':'aryan@hdfc.com','age': 30, 'weight': 75.2, 'height': 3} # if you will right str instead of int in age it will raise error
patient1 = Patient(**patient_info) #validation - type coercion pefrome in this step


#step 3:
insert_patient_data(patient1)
```
