```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Data/wine_data.csv')
```

```python
# df.columns=['Class label', 'Alcohol', 'Malic acid']
df = df.iloc[:, :3]
```

```python
df
```

```python
df.columns = ['Class', 'Alcohol', 'Malic_Acid']
```

```python
df
```

```python
sns.kdeplot(x=df['Alcohol'])
```

```python
sns.kdeplot(x=df['Malic_Acid'])
```

```python
color_dict = {1:'red', 3:'green', 2:'blue'}
sns.scatterplot(x=df['Alcohol'], y=df['Malic_Acid'], hue=df['Class'], palette=color_dict)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop('Class', axis=1),df['Class'],test_size=0.3,random_state=0)

X_train.shape, X_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X_test.columns)
```

```python
np.round(X_train.describe(),1)
```

```python
np.round(X_train_scaled.describe(),1)
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

ax1.scatter(X_train['Alcohol'],
            X_train['Malic_Acid'],
            c=y_train)

ax1.set_title("Before Scaling")

ax2.scatter(X_train_scaled['Alcohol'],
            X_train_scaled['Malic_Acid'],
            c=y_train)

ax2.set_title("After Scaling")

plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# Before Scaling
ax1.set_title('Before Scaling')
sns.kdeplot(X_train['Alcohol'], ax=ax1)
sns.kdeplot(X_train['Malic_Acid'], ax=ax1)

# After Scaling
ax2.set_title('After Standard Scaling')
sns.kdeplot(X_train_scaled['Alcohol'], ax=ax2)
sns.kdeplot(X_train_scaled['Malic_Acid'], ax=ax2)

plt.show()
```
