## Nested Models in Pydantic

Nested models mean:

- one Pydantic model inside another model.

- Used when data is complex and contains sub-objects.

### Suppose we store the entire address as a single string.
- Then extracting fields like pincode, city, or area name becomes difficult and messy.

- Nested models help us store address data in a structured format,
- making validation and accessing individual fields much easier.

```python
from pydantic import BaseModel

#step 1 : model
class Address(BaseModel):
    city: str
    state: str
    pin: int

class Patient(BaseModel):

    name: str
    gender: str
    age : int
    address: Address # address is complex data 


def detail(patient: Patient):
    print(patient.name)
    print(patient.gender)
    print(patient.age)
    print(patient.address)
    

    
# step 2 : dict

address_dict = {'city':'amd', 'state':'guj', 'pin': 646146}
address1 = Address(**address_dict)

patient_dict = {'name': 'Aryan', 'gender':'male', 'age': 22, 'address': address1}
patient1 = Patient(**patient_dict)


detail(patient1)
```

## Benefit	Explanation

- Better structure -->	Clean organized code
- Reusability --> Same child model can be reused
- Better validation -->	Validation on nested fields
- Real-world modeling --> Matches API/JSON structure
