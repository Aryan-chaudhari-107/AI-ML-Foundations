## Import Data & plot graph

```python
# from a pandas dataframe
batsman  =pd.read_csv('Data/sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'], batsman['V Kohli'])
```

## plotting multiple plots

```python

plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
```

## labels title

```python
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])

plt.title('Rohit Sharma VS Virat Kohli Career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
```

## color(hex) 
## line(width and style) 
## marker(size)

```python
plt.plot(batsman['index'], batsman['V Kohli'], color='green')
plt.plot(batsman['index'], batsman['RG Sharma'], color='blue')

plt.title('Rohit Sharma VS Virat Kohli Career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='dashed', linewidth=15)
plt.plot(batsman['index'], batsman['RG Sharma'], color='blue', linestyle='dotted', linewidth=7)

plt.title('Rohit Sharma VS Virat Kohli Career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], marker='D', markersize=15)
plt.plot(batsman['index'], batsman['RG Sharma'], marker='+', markersize= 5)

plt.title('Rohit Sharma VS Virat Kohli Career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored') 
```

## legend -> location

```python

plt.plot(batsman['index'], batsman['V Kohli'], color='green', label='Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color='blue', label='Rohit')

plt.title('Rohit Sharma VS Virat Kohli Career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')

plt.legend() 
# plt.legend(loc='right') for particular location
# do not forget legend line 
```

## Outliers Effect & Solution

```python

price = [48000,54000,57000,60000,65000,42000, 4500000]
year = [2015, 2016, 2017, 2018, 2019, 2020, 2021]

plt.plot(year,price) 
```

```python
# limiting axes (solution)
price = [48000,54000,57000,60000,65000,42000, 4500000]
year = [2015, 2016, 2017, 2018, 2019, 2020, 2021]

plt.plot(year,price) 
plt.ylim(0,75000)
plt.xlim(2017,2019)
```

## grid

```python
# grid
plt.plot(batsman['index'], batsman['V Kohli'], color='green', label='Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color='blue', label='Rohit')

plt.title('Rohit Sharma VS Virat Kohli Career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
plt.legend() 

plt.grid()
```

## plt.show()

- Use plt.show() in Python scripts (.py) because Matplotlib creates the plot but doesn't automatically display it.
- Don't usually need it in Jupyter Notebook (.ipynb) or Colab because these environments automatically render the last plot output of a cell.

```python

plt.plot(batsman['index'], batsman['V Kohli'], color='green', label='Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color='blue', label='Rohit')

plt.title('Rohit Sharma VS Virat Kohli Career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
plt.legend() 

plt.grid()

plt.show() 
```
