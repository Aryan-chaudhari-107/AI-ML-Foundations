```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Bar Chart


- Bivariate Analysis
- Numerical vs Categorical
- Use case: Aggregate analysis of groups


## simple bar chart

```python
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children,color='black')
```

# horizontal bar chart

```python
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.barh(colors,children,color='black')
```

## color and label

```python
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.barh(colors,children,color='gold')
plt.xlabel('childern')
plt.ylabel('color')
```

```python
df = pd.read_csv('Data/batsman_season_record.csv')
```

```python
df
```

```python
plt.bar(df['batsman'], df['2015'])
```

```python
plt.bar(np.arange(df.shape[0]) - 0.3, df['2015'], width=0.3, color='yellow')
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.3, color='red')
plt.bar(np.arange(df.shape[0]) + 0.3, df['2017'], width=0.3, color='blue')

plt.xticks(np.arange(df.shape[0]), df['batsman'])

plt.show()
```

## A Problem

```python
# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='black')
# plt.xticks(rotation='vertical')
plt.show()
```

```python
# solution
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical')
plt.show()
```

## Stacked Bar Chart

```python
plt.bar(df['batsman'], df['2017'], label='2017')
plt.bar(df['batsman'], df['2016'], bottom=df['2017'], label='2016')
plt.bar(df['batsman'], df['2015'], bottom=df['2016']+df['2017'], label='2015')

plt.legend()
plt.show()
```
