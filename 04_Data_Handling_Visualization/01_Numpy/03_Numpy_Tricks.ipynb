## Set functions

- np.union1d
- np.intersect1d
- np.setdiff1d
- np.setxor1d
- np.in1d

NumPy set functions are useful for finding unique/common/different values between arrays quickly and easily.

```python
m = np.array([1,2,3,4,5])
n = np.array([3,4,5,6,7])

np.union1d(m,n)
```

```python
np.intersect1d(m, n)
```

```python
np.setdiff1d(m,n)
```

```python
np.setxor1d(m,n) # remove common
```

```python
np.in1d(m,n)
```

## np.allclose

Checks if arrays are almost equal.

Useful for decimal/floating point comparisons.

```python

a = np.array([1.00001,2.00001])
b = np.array([1.00002,2.00002])

print(np.allclose(a,b))
```
