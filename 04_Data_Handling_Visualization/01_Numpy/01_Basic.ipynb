## What Is NumPy??

- NumPy stands for Numerical Python. It is a Python library used for fast numerical and mathematical computations. NumPy is mainly used in Data Science, Machine Learning, and scientific computing.

- The core feature of NumPy is the ndarray (N-dimensional array). An ndarray stores data in the form of arrays and allows fast operations on large amounts of data.

- NumPy arrays are faster and more memory efficient than normal Python lists because they store elements of the same data type together.

- With NumPy, we can perform:

-- Mathematical operations
-- Statistical calculations
-- Matrix operations
-- Array manipulation
-- Random number generation
-- Linear algebra operations

- NumPy is widely used because many libraries like Pandas, Scikit-learn, TensorFlow, and PyTorch are built on top of it.

- In simple words, NumPy is a fast and powerful library that helps Python work efficiently with numerical data and arrays.

## NumPy Arrays VS Python Lists.

- NumPy arrays and Python lists are used to store data, but they work differently.

- Python lists are flexible. We can add or remove items anytime, and we can store different types of data together like numbers, strings, and floats.

- NumPy arrays are more strict. After creating an array, its size is fixed. Also, NumPy arrays usually store only one type of data, like all integers or all floats.

- Because NumPy stores the same type of data together, it becomes much faster and uses less memory than Python lists.

- NumPy is specially made for mathematical and numerical calculations. It can handle large amounts of data very quickly and with less code.

- That is why NumPy is widely used in Data Science, Machine Learning, and scientific computing. Many libraries like Pandas and TensorFlow use NumPy internally.

- In simple words:

Python lists are flexible and easy to use.
NumPy arrays are faster and better for calculations.

## Creating NumPy Arrays.

```python
#--> 1: np.array 

#1D (also know as a vector)
import numpy as np

a = np.array([1,2,3])
print(a)

print(type(a))
```

```python
#2D (also know as a metrix )

b = np.array([
    [1,2,3],[4,5,6]
    ])
    
print(b)
print(type(b))
```

```python
#3D (also know as a tensor)

c = np.array([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
    ])
    
print(c)
```

```python
#--> 2: dtype (Create Array Of Your Choice)

x = np.array([1,2,3], dtype=float)
y = np.array([1,2,3], dtype=bool)
z = np.array([1,2,3], dtype=complex)

print(x)
print(y)
print(z)
```

```python
#--> 3: np.arange (use for create array)

p= np.arange(1,11)
q = np.arange(1,11,2)
r = np.arange(1,11, dtype=float)

print(p)
print(q)
print(r)
```

```python
#--> 4: reshape

aa = np.arange(1,11).reshape(5,2) #reshape(row, column) 
bb = np.arange(1,11).reshape(2,5) 

print(aa)
print("\n")
print(bb)
```

```python
#--> 5: np.ones and np.zeros and np.random

m = np.ones((3,4)) #pass tuple not array (row, column)
n = np.zeros((5,2)) 
o = np.random.random((3,2)) #give numben b/w 0 to 1

print(m)
print("\n")
print(n)
print("\n")
print(o)
```

```python
#--> 6: np.linspace (linear space)
#generate points at equal distance 

np.linspace(-10, 10, 10) #(lowerrange, upperrange, num of items)

```

```python
#--> 7: np.identity 

np.identity(3) #diagnoal items are 1 and others are 0 
```

## Indexing and Slicing

```python
a5 = np.arange(10)
a6 = np.arange(12).reshape(3,4)
a7 = np.arange(27).reshape(3,3,3)

print(a5)
print(a6)
print(a7)
```

```python
#--> 1: Indexing 

# In 1D 
a5[-1]  # positiven, negative same as python
```

```python
# In 2D 

a6[1,2] #(row, column)
```

```python
# In 3D 

a7[1,0,1] #(which 2d array, row, column)
```

```python
#--> 2: slicing 

# 1D

a5[2:10:2] #(same as python (start, end, skip))
```

```python
#2D 
 
# FORMAT: (row, column) #ask yourself whick row and which column 
# if you want continus row and column used single --> :
# if you want skip any row and column use double --> ::

a6[0,:] #(first row, {:} all column)

a6[:,2] # (all row, 2nd column)

a6[1:, 1:3] 

a6[::2,::3]

a6[::2, 1::2]

a6[1,0::3]

a6[0:2, 1::]
```

```python

#3D 

#FORMAT --> (which 2D array, row, column)
# just give which 2D array and then same login for row and column as per 2D array
# at the end 3D array is made form 2D array so same logic 

a7[1] # for middle 2d array

a7[0::2] #i want first and last 2d array from 3d array

a7[0,1] #first 2d array 2nd row

a7[1,:,1] #2nd 2d array ka middle column

a7[2,1:,1:] # i want [22,23],[25,26]

a7[::2,0,0::2]
```

## Stacking

[2 X 2] [2 X 2] these are the arrays

when we do staking:

horizontal : [2 X 4]
vertical : [4 X 2]

you can do with n number of arrays
no limitation

Rule: shape must be same of all arrays

```python
a15 = np.arange(12).reshape(3,4)
a16 = np.arange(12,24).reshape(3,4)

print(a15)
print(a16)
```

```python
#--> 1: horizontal stacking

np.hstack((a15,a16)) #np.hstack((1,2,3,...,n))

```

```python
#--> 2: vertical stacking

np.vstack((a15,a16))
```
