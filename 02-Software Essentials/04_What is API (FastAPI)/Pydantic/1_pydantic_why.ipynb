## POOR WAY:

- Problem with poor way.

```python
# Python has Dynamic Typing

# Problem:
# Python does not strictly enforce data types at runtime.

# Senior developer created this function
def insert_patient_data(name: str, age: int):

    print(f"Name : {name}")
    print(f"Age  : {age}")

    print("Inserted into database")


# Junior developer uses the function incorrectly
insert_patient_data("Aryan", "Thirty")


# Output:
# Name : Aryan
# Age  : Thirty
# Inserted into database


# ❌ Problem:
# Even though age should be int,
# Python still accepts string values.

# Type hints only help developers/readability.
# They DO NOT raise errors automatically.
```

- This is not scalable because for every function,

```python
# Better Example with Validation

def insert_patient_data(name: str, age: int):

    if type(name) == str and type(age) == int:
        if age < 0:
            raise ValueError('Age cant be negative')

        else:
            print(f"Name : {name}")
            print(f"Age  : {age}")
            print("Inserted into database")
    
    else:
        raise TypeError('Incorrect data')
    


# Correct
insert_patient_data("Aryan", 21)

# Wrong
insert_patient_data("Aryan", "Twenty One")
# Raises TypeError


# if i create
def update_patient_data(name:str, age: int):
# i have to write same logic for this as well
```

## USING PYDANTIC:

- Pydantic handles type validation automatically

# Pydantic Workflow

### 1. Create a Pydantic model
- Define fields
- Define data types
- Add validations if needed

### 2. Pass raw input data   
- Dictionary / JSON data

### 3. Pydantic automatically:
- Validates data
- Converts compatible types
- Raises ValidationError if data is invalid

### 4. Use validated data safely in your application


Result:
- Clean, type-safe, validated data across the project

```python
from pydantic import BaseModel


# step 1: model
class Patient(BaseModel):
    name: str
    age: int


def insert_patient_data(patient: Patient):
    print(patient.name)
    print(patient.age)
    print("Inserted into database")

def updated_patient_data(patient: Patient):
    print(patient.name)
    print(patient.age)
    print("Inserted into database")


#step 2: pass raw input data (created dict)

patient_info = {'name': 'aryan', 'age':30} # if you will right str instead of int in age it will raise error
patient1 = Patient(**patient_info) #unpack dict


#step 3:
insert_patient_data(patient1)
```
