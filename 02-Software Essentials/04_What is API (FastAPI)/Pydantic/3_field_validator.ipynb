## What is mode in Pydantic?

- mode decides WHEN the validator should run.

- @field_validator('field_name', mode='before')

- @field_validator('field_name', mode='after')

```python
from pydantic import BaseModel, EmailStr, AnyUrl, Field, field_validator
from typing import List,Dict, Optional, Annotated

# step 1: model
class Patient(BaseModel):

    name: str
    email: EmailStr
    age: int
    weight: float 
    married: bool
    allergies: List[str] 
    contact_detials: Dict[str, str] 
    address : Optional[str] = None 
    
    #-------------------------------------------------------------------
    # VERIFY:
    #-------------------------------------------------------------------
    @field_validator('email')   # Validate the email field
    @classmethod                # Makes this method a class method
    def email_validator(cls, value):  # cls = class, value = email input

        valid_domains = ['hdfc.com', 'icici.com']

        # abc@gmail.com -> gmail.com
        domain_name = value.split('@')[-1]

        if domain_name not in valid_domains:
            raise ValueError('Not a valid domain')

        return value
    #-------------------------------------------------------------------




    #------------------------------------------------------------------
    # TRANSFORM:
    #------------------------------------------------------------------
    @field_validator('name')
    @classmethod
    def transform_name(cls, value):
        return value.upper()
    #------------------------------------------------------------------
    




    #------------------------------------------------------------------
    # MODE:
    #------------------------------------------------------------------
    @field_validator('age', mode='before')
    @classmethod
    def validate_age(cls, value):

        if 0 < value < 100:
            return value
        else:
            raise ValueError('Age should be b/w 0 to 100')
     #------------------------------------------------------------------




def insert_patient_data(patient: Patient):
    print(patient.name)
    print(patient.email)
    print(patient.age)
    print("Inserted into database")


#step 2: pass raw input data (created dict)

patient_info = {'name': 'aryan', 'email':'aryan@hdfc.com','age': '30'}
patient1 = Patient(**patient_info) #validation - type coercion pefrome in this step


#step 3:
insert_patient_data(patient1)
```

```python
# -------------------- mode='before' --------------------
# Runs BEFORE Pydantic performs type conversion (type coercion)

@field_validator('age', mode='before')

# In your input:
# 'age': '30'

# Here value will be:
# value = '30'   -> string

# Because validation runs before conversion,
# the validator receives raw input data directly.

# After validation:
# Pydantic converts '30' -> 30 (int)



# -------------------- mode='after' --------------------
# Runs AFTER Pydantic performs type conversion

@field_validator('age', mode='after')

# In your input:
# 'age': '30'

# First Pydantic converts:
# '30' -> 30

# Then validator runs.

# Here value will be:
# value = 30   -> integer

# So after mode gets already converted/validated data.



# -------------------- SUMMARY --------------------

# mode='before'
# Raw Input  -> Validator -> Type Conversion

# mode='after'
# Raw Input -> Type Conversion -> Validator
```
